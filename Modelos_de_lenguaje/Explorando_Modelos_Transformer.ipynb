```python
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline

# Cargar modelo y tokenizador
tokenizer = AutoTokenizer.from_pretrained("microsoft/Phi-3-mini-4k-instruct")

model = AutoModelForCausalLM.from_pretrained(
    "microsoft/Phi-3-mini-4k-instruct",
    device_map="cuda",
    torch_dtype="auto",
    trust_remote_code=False,
)

# Crear un pipeline de generación de texto
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    return_full_text=False,
    max_new_tokens=50,
    do_sample=False,
)
```

```python
# Entradas y Salidas de un Modelo Transformer Entrenado

prompt = "Escribe un mensaje disculpándote con Juan por haber pisado su pie en el colectivo. Explica cómo sucedió."

output = generator(prompt)
print(output[0]['generated_text'])
```

```python
# Mostrar la arquitectura del modelo
print(model)
```

```python
# Seleccionando un token de la distribución de probabilidades

prompt = "La capital de Argentina es"

input_ids = tokenizer(prompt, return_tensors="pt").input_ids.to("cuda")

model_output = model.model(input_ids)
lm_head_output = model.lm_head(model_output[0])

token_id = lm_head_output[0, -1].argmax(-1)
print(tokenizer.decode(token_id))
```

```python
# Acelerando la generación con y sin caché

prompt = "Escribe una carta muy larga explicando por qué el asado del domingo se quemó. Describe qué salió mal."

input_ids = tokenizer(prompt, return_tensors="pt").input_ids.to("cuda")

%%timeit -n 1
generation_output = model.generate(
  input_ids=input_ids,
  max_new_tokens=100,
  use_cache=True
)

%%timeit -n 1
generation_output = model.generate(
  input_ids=input_ids,
  max_new_tokens=100,
  use_cache=False
)
```
